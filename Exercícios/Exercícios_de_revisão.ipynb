```python
#Exercício 01: Crie um programa que leia 10 números inteiros e armazene-os em um array.

vetor = [] #criando um vetor para armazenar os números
soma = 0

for i in range(10): #criando um for para gerar um loop de 10 números inseridos
    numero = int(input(f"Digite o {i + 1}º número: "))
    vetor.append(numero)
    soma += numero

media = soma / 10

print("\nNúmeros digitados:") #mostrando os números escolhidos
for numero in vetor:
    print(numero, end=" ")

print(f"\nSoma: {soma}") #mostrando as soma dos números inseridos
print(f"Média: {media:.2f}")
```

```python
#Exercício 02: Leia 10 números inteiros utilizando um array.

vetor = [] #criando um vetor para armazenar os números

for i in range(10): #criando um for para gerar um loop de 10 números inseridos
    numero = int(input(f"Digite o {i + 1}º número: "))
    vetor.append(numero)

maior = vetor[0]
menor = vetor[0]

for numero in vetor: #um for para mostrar o maior e o menor número
    if numero > maior:
        maior = numero

    if numero < menor:
        menor = numero

print(f"\nMaior valor inserido: {maior}")
print("Posições do maior:", end=" ")

for i in range(10):
    if vetor[i] == maior:
        print(i, end=" ")

print(f"\nMenor valor inserido: {menor}")
print("Posições do menor:", end=" ")

for i in range(10):
    if vetor[i] == menor:
        print(i, end=" ")

print()
```

```python
#Exercício 03: Leia 20 números inteiros e armazene-os em um array.

vetor = []

for i in range(20): #criando um for para gerar um loop de 10 números inseridos
    numero = int(input(f"Digite o {i + 1}º número: "))
    vetor.append(numero)

quantidade = 0
soma = 0

print("\nNúmeros pares inseridos:")

for numero in vetor:
    if numero % 2 == 0:
        print(numero, end=" ")
        quantidade += 1
        soma += numero

print(f"\n\nQuantidade de números pares: {quantidade}")
print(f"Soma dos números pares: {soma}")
```

```python
#Exercício 04: Leia 10 números e armazene-os em um array.

vetor = []

for i in range(10):
    numero = int(input(f"Digite o {i + 1}º número: "))
    vetor.append(numero)

print("\nVetor original:")
print(vetor)

# Invertendo o próprio vetor
inicio = 0
fim = 9

while inicio < fim:
    aux = vetor[inicio]
    vetor[inicio] = vetor[fim]
    vetor[fim] = aux

    inicio += 1
    fim -= 1

print("\nVetor invertido:")
print(vetor)
```

```python
#Exercicio 05: Crie uma matriz 3 x 3 de números inteiros.

matriz = []

# Leitura da matriz
for i in range(3):
    linha = []

    for j in range(3):
        numero = int(input(f"Digite o valor [{i}][{j}]: "))
        linha.append(numero)

    matriz.append(linha)

soma = 0
maior = matriz[0][0]

print("\nMatriz:")

for i in range(3):
    for j in range(3):
        print(matriz[i][j], end=" ")

        soma += matriz[i][j]

        if matriz[i][j] > maior:
            maior = matriz[i][j]

    print()

print(f"\nSoma de todos os elementos: {soma}")
print(f"Maior valor: {maior}")
```

```python
#Exercicio 06: Leia uma matriz 4 x 4. Mostre os elementos da diagonal principal e calcule sua soma.

matriz = []

# Criando a matriz 4x4
for i in range(4):
    linha = []

    for j in range(4):
        numero = int(input(f"Digite o valor [{i}][{j}]: "))
        linha.append(numero)

    matriz.append(linha)

soma = 0

print("\nDiagonal principal:")

for i in range(4):
    print(matriz[i][i], end=" ")
    soma += matriz[i][i]

print(f"\n\nSoma: {soma}")
```

```python
#Exercicio 07: Uma turma possui 4 alunos e 3 avaliações.]

notas = []

# Cadastro das notas
for i in range(4):
    notas_aluno = []

    print(f"\nAluno {i + 1}")

    for j in range(3):
        nota = float(input(f"Digite a nota P{j + 1}: "))
        notas_aluno.append(nota)

    notas.append(notas_aluno)

# Calculando as médias
print("\nMédias dos alunos:")

for i in range(4):
    soma = 0

    for j in range(3):
        soma += notas[i][j]

    media = soma / 3

    print(f"Aluno {i + 1}: {media:.2f}")
```

```python
#Exercício 08: Crie uma struct Produto contendo:

class Produto:
    def __init__(self, nome, codigo, preco, quantidade):
        self.nome = nome
        self.codigo = codigo
        self.preco = preco
        self.quantidade = quantidade

    def valor_estoque(self):
        return self.preco * self.quantidade


produtos = []

# Cadastro dos produtos
for i in range(5):
    print(f"\nProduto {i + 1}")

    nome = input("Nome: ")
    codigo = int(input("Código: "))
    preco = float(input("Preço: "))
    quantidade = int(input("Quantidade: "))

    produto = Produto(nome, codigo, preco, quantidade)

    produtos.append(produto)

# Exibição
maior_valor = 0
produto_maior = None

print("\n========== PRODUTOS ==========")

for produto in produtos:
    valor = produto.valor_estoque()

    print(f"\nNome: {produto.nome}")
    print(f"Código: {produto.codigo}")
    print(f"Preço: R$ {produto.preco:.2f}")
    print(f"Quantidade: {produto.quantidade}")
    print(f"Valor em estoque: R$ {valor:.2f}")

    if valor > maior_valor:
        maior_valor = valor
        produto_maior = produto

print("\n==============================")
print("Produto com maior valor em estoque:")
print(f"Nome: {produto_maior.nome}")
print(f"Valor: R$ {maior_valor:.2f}")
```

```python
#exercicio 09: Crie uma struct Aluno contendo:

class Aluno:
    def __init__(self, nome, idade, nota1, nota2, nota3):
        self.nome = nome
        self.idade = idade
        self.nota1 = nota1
        self.nota2 = nota2
        self.nota3 = nota3

    def media(self):
        return (self.nota1 + self.nota2 + self.nota3) / 3


alunos = []

# Cadastro
for i in range(5):
    print(f"\nAluno {i + 1}")

    nome = input("Nome: ")
    idade = int(input("Idade: "))
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))

    aluno = Aluno(nome, idade, nota1, nota2, nota3)

    alunos.append(aluno)


aprovados = 0
reprovados = 0

maior_media = 0
aluno_maior_media = None

print("\n========== RESULTADO ==========")

for aluno in alunos:

    media = aluno.media()

    if media >= 7:
        situacao = "Aprovado"
        aprovados += 1
    else:
        situacao = "Reprovado"
        reprovados += 1

    print(f"{aluno.nome} - Média: {media:.2f} - {situacao}")

    if media > maior_media:
        maior_media = media
        aluno_maior_media = aluno


print("\n===============================")
print(f"Aprovados: {aprovados}")
print(f"Reprovados: {reprovados}")

print(f"\nAluno com maior média: {aluno_maior_media.nome}")
print(f"Maior média: {maior_media:.2f}")
```

```python
#Exercicio 10: Crie um pequeno sistema para cadastro de 10 funcionários.

class Funcionario:
    def __init__(self, nome, idade, cargo, salario):
        self.nome = nome
        self.idade = idade
        self.cargo = cargo
        self.salario = salario


funcionarios = []


def cadastrar_funcionarios():
    if len(funcionarios) >= 10:
        print("\nJá existem 10 funcionários cadastrados!")
        return

    quantidade = 10 - len(funcionarios)

    print(f"\nVocê pode cadastrar até {quantidade} funcionário(s).")

    while len(funcionarios) < 10:

        print(f"\nFuncionário {len(funcionarios) + 1}")

        nome = input("Nome: ")
        idade = int(input("Idade: "))
        cargo = input("Cargo: ")
        salario = float(input("Salário: "))

        funcionario = Funcionario(
            nome,
            idade,
            cargo,
            salario
        )

        funcionarios.append(funcionario)

        continuar = input("\nDeseja cadastrar outro? (s/n): ")

        if continuar.lower() != "s":
            break


def listar_funcionarios():

    if len(funcionarios) == 0:
        print("\nNenhum funcionário cadastrado.")
        return

    print("\n========== FUNCIONÁRIOS ==========")

    for funcionario in funcionarios:

        print(f"\nNome: {funcionario.nome}")
        print(f"Idade: {funcionario.idade}")
        print(f"Cargo: {funcionario.cargo}")
        print(f"Salário: R$ {funcionario.salario:.2f}")


def maior_salario():

    if len(funcionarios) == 0:
        print("\nNenhum funcionário cadastrado.")
        return

    maior = funcionarios[0]

    for funcionario in funcionarios:

        if funcionario.salario > maior.salario:
            maior = funcionario

    print("\n========== MAIOR SALÁRIO ==========")
    print(f"Funcionário: {maior.nome}")
    print(f"Cargo: {maior.cargo}")
    print(f"Salário: R$ {maior.salario:.2f}")


def media_salarial():

    if len(funcionarios) == 0:
        print("\nNenhum funcionário cadastrado.")
        return

    soma = 0

    for funcionario in funcionarios:
        soma += funcionario.salario

    media = soma / len(funcionarios)

    print("\n========== MÉDIA SALARIAL ==========")
    print(f"Média salarial: R$ {media:.2f}")


def salarios_acima_media():

    if len(funcionarios) == 0:
        print("\nNenhum funcionário cadastrado.")
        return

    soma = 0

    for funcionario in funcionarios:
        soma += funcionario.salario

    media = soma / len(funcionarios)

    print(f"\nFuncionários com salário acima da média (R$ {media:.2f}):")

    encontrou = False

    for funcionario in funcionarios:

        if funcionario.salario > media:
            print(
                f"{funcionario.nome} - "
                f"R$ {funcionario.salario:.2f}"
            )

            encontrou = True

    if not encontrou:
        print("Nenhum funcionário possui salário acima da média.")


# MENU PRINCIPAL

while True:

    print("\n=================================")
    print("      SISTEMA DE FUNCIONÁRIOS")
    print("=================================")
    print("1 - Cadastrar funcionários")
    print("2 - Listar funcionários")
    print("3 - Maior salário")
    print("4 - Média salarial")
    print("5 - Salários acima da média")
    print("0 - Sair")
    print("=================================")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        cadastrar_funcionarios()

    elif opcao == "2":
        listar_funcionarios()

    elif opcao == "3":
        maior_salario()

    elif opcao == "4":
        media_salarial()

    elif opcao == "5":
        salarios_acima_media()

    elif opcao == "0":
        print("\nPrograma encerrado!")
        break

    else:
        print("\nOpção inválida!")
```
