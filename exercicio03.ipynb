```python
#Exercício 01: Crie um programa que leia 10 números inteiros e armazene-os em um array.

vetor = [] #criando um vetor para armazenar os números
soma = 0

for i in range(10): #criando um for para gerar um loop de 10 números inseridos
    numero = int(input(f"Digite o {i + 1}º número: "))
    vetor.append(numero)
    soma += numero

media = soma / 10

print("\nNúmeros digitados:") #mostrando os números escolhidos
for numero in vetor:
    print(numero, end=" ")

print(f"\nSoma: {soma}") #mostrando as soma dos números inseridos
print(f"Média: {media:.2f}")
```

```python
#Exercício 03: Leia 20 números inteiros e armazene-os em um array.

vetor = []

for i in range(20): #criando um for para gerar um loop de 10 números inseridos
    numero = int(input(f"Digite o {i + 1}º número: "))
    vetor.append(numero)

quantidade = 0
soma = 0

print("\nNúmeros pares inseridos:")

for numero in vetor:
    if numero % 2 == 0:
        print(numero, end=" ")
        quantidade += 1
        soma += numero

print(f"\n\nQuantidade de números pares: {quantidade}")
print(f"Soma dos números pares: {soma}")
```
